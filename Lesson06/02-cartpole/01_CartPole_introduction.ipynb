```python
import gymnasium as gym
```

```python
# Create the CartPole environment with human rendering mode
env = gym.make("CartPole-v1") #, render_mode="human")
```

## Observation Space

The observation is an `ndarray` with shape `(4,)`, with values corresponding to the following positions and velocities:

| Index | Observation | Minimum | Maximum |
|---:|---|---:|---:|
| 0 | Cart position | -4.8 | 4.8 |
| 1 | Cart velocity | -∞ | ∞ |
| 2 | Pole angle | ≈ -0.418 rad (-24°) | ≈ 0.418 rad (24°) |
| 3 | Pole angular velocity | -∞ | ∞ |

These bounds describe the possible observations, but an episode terminates before the full bounds are reached: the cart position must remain within `(-2.4, 2.4)`, and the pole angle must remain within approximately `(-0.2095, 0.2095)` radians (±12°).

## Rewards

By default, the environment gives a reward of `+1` for every step, including the termination step. The default reward threshold for `CartPole-v1` is `500`.

## Starting State

At the start of an episode, all four observation values are sampled uniformly from `(-0.05, 0.05)`.

## Episode End

An episode ends when any of the following occurs:

- **Termination:** the pole angle exceeds ±12°.
- **Termination:** the cart position exceeds ±2.4.
- **Truncation:** the episode length exceeds 500 steps for `CartPole-v1`.

Source: [Gymnasium Cart Pole documentation](https://gymnasium.farama.org/environments/classic_control/cart_pole/).


```python
# Reset the environment to start a new episode
state, _ = env.reset()
print("Initial state:", state)
```

```python
print("Max episode steps:", env.spec.max_episode_steps)
print("Reward threshold:", env.spec.reward_threshold)
print("State space:", env.observation_space)
print("Action space:", env.action_space)
```

```python
# Initial state: [cart position, cart velocity, pole angle, pole angular velocity]
print("cart position.  :", state[0]) 
print("cart velocity   :", state[1])
print("pole angle      :", state[2])
print("pole angular vel:", state[3])
```

```python
# Take an action (0 or 1) to see how the environment responds
print("cart position.  :", state[0]) 
print("cart velocity   :", state[1])
print("pole angle      :", state[2])
print("pole angular vel:", state[3])
print(30*"-")
action = 0  # 0 push cart to the left / 1 push cart to the right
ACTION_SYMBOLS = {0: "left", 1: "right"}
print(f"Taking action {action} (push cart to the {ACTION_SYMBOLS[action]})...")
state, reward, terminated, truncated, info = env.step(action)
print("cart position.  :", state[0]) 
print("cart velocity   :", state[1])
print("pole angle      :", state[2])
print("pole angular vel:", state[3])
```
