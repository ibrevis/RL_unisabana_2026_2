```python
import gymnasium as gym
```

```python
# Create the FrozenLake environment with human rendering mode
env = gym.make("FrozenLake-v1",
               map_name="4x4",
               is_slippery=False,
               reward_schedule=(1, -1, 0),   # (goal, hole, step)
               render_mode="ansi"
               )
```

## Description

FrozenLake is a grid-world environment in which the player must travel from the start tile to the goal without falling into a hole. The predefined 4×4 map starts at location `[0, 0]` and places the goal at `[3, 3]`. An episode continues until the player reaches the goal or enters a hole.

The lake is slippery by default, which can make the player move perpendicular to the chosen direction. This notebook sets `is_slippery=False`, so each action moves in its intended direction.

## Action Space

The action space is `Discrete(4)`, with one integer for each direction:

| Action | Direction |
|---:|---|
| 0 | Left |
| 1 | Down |
| 2 | Right |
| 3 | Up |

## Observation Space

The observation is a single integer representing the player's position. It is calculated as `current_row * number_of_columns + current_column`, with row and column numbering starting at 0. The 4×4 map therefore has observation space `Discrete(16)`, containing states 0 through 15. For example, the goal at `[3, 3]` is state `3 * 4 + 3 = 15`.

## Starting State

Each episode begins in state `0`, corresponding to location `[0, 0]`.

## Rewards

By default, reaching the goal gives `+1`, while entering a hole or moving onto a frozen tile gives `0`. This notebook uses `reward_schedule=(1, -1, 0)`, so the goal gives `+1`, a hole gives `-1`, and a frozen tile gives `0`.

## Episode End

An episode terminates when the player enters a hole or reaches the goal. For the 4×4 environment, the time-limit wrapper truncates an episode after 100 steps.

Source: [Gymnasium FrozenLake documentation](https://gymnasium.farama.org/environments/toy_text/frozen_lake/).


```python
# Reset the environment to start a new episode
state, _ = env.reset()
print("Initial state:", state)
```

```python
print("Max episode steps:", env.spec.max_episode_steps)
print("Reward threshold:", env.spec.reward_threshold)
print("State space:", env.observation_space)
print("Action space:", env.action_space)
```

```python
# Initial state: 
print("Initial state:", state) 
print(env.render())  # Render the initial state
```

```python
# Take an action (0, 1, 2, 3) to see how the environment responds
print(env.render())
print(30*"-")
action = 1  # 0 left, 1 down, 2 right, 3 up
ACTION_SYMBOLS = {0: "left", 1: "down", 2: "right", 3: "up"}
print(f"Taking action {action} ({ACTION_SYMBOLS[action]})...")
state, reward, terminated, truncated, info = env.step(action)
print("State:", state) 
print("Reward:", reward)
print("Terminated:", terminated)
print("Truncated:", truncated)
print(env.render())  # Render the new state
```

```python
env.close()
```
